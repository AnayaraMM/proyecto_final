```python
import requests
import pandas as pd
import time
```

```python
# Obtenemos Segovia por separado ya que el enlace es diferente
years = [2024, 2025]
meses = ["enero","febrero","marzo","abril", "mayo", "junio", "julio", "agosto", "septiembre", "octubre", "noviembre", "diciembre"]
mes_dict = {"enero":1,"febrero":2,"marzo":3,"abril":4,"mayo":5,"junio":6,"julio":7,"agosto":8,"septiembre":9,"octubre":10,"noviembre":11,"diciembre":12}

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}
base_url = "https://www.tutiempo.net/estaciones/segovia/{}-{}/"

dfs = []
errors = []
for year in years:
  for mes in meses:
      url = base_url.format(mes, year)
      print(f"Descargando {url}...")

      try:
        response = requests.get(url, headers=headers)
        response.encoding = response.apparent_encoding

        tables = pd.read_html(response.text)
        df_month = tables[0]

        df_month["Mes"] = mes_dict[mes]
        df_month["Year"] = str(year)
        df_month["Provincia"] = "Segovia"

        dfs.append(df_month)

        time.sleep(2)

      except Exception as e:
        errors.append(f"Error en {year}: {mes}")


print(errors)
df_segovia = pd.concat(dfs, ignore_index=True)
df_segovia.columns = df_segovia.columns.str.strip()
```

```python
df_segovia["fecha"] = df_segovia["Día"].astype(str).str.zfill(2) + "/" + df_segovia["Mes"].astype(str).str.zfill(2) + "/" + df_segovia["Year"].astype(str)
```

```python
# Meses a scrapear
years = [2024, 2025]
provincias = [{"Ávila":"80110"},{"Burgos":"80750"},{"León":"80550"},{"Salamanca":"82020"}, {"Soria":"81480"}, {"Valladolid":"81410"}, {"Zamora":"81300"}]
# Ávila, Burgos, León, Salamanca, Soria, Valladolid, Zamora

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}
base_url = "https://www.tutiempo.net/clima/{}/ws-{}.html"

dfs = []
errors = []
for year in years:
  months = [f"{month:02d}-{year}" for month in range(1, 13)]
  for provincia in provincias:
    for clave, valor in provincia.items():
      for month in months:
          url = base_url.format(month, valor)
          print(f"Descargando {url}...")

          try:
            response = requests.get(url, headers=headers)
            response.encoding = response.apparent_encoding

            tables = pd.read_html(response.text)
            df_month = tables[0]

            df_month["Mes"] = int(month.split('-')[0])
            df_month["Year"] = year
            df_month["Provincia"] = clave

            dfs.append(df_month)

            time.sleep(2)

          except Exception as e:
            errors.append(f"Error en {clave}: {month} - {year}")


print(errors)
# Unir todos los meses
final_df = pd.concat(dfs, ignore_index=True)
final_df.columns = final_df.columns.str.strip()
```
